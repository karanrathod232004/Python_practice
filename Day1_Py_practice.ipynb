#Arithmetic operators

```python
salary = 40000
bonus = 5000
Total_salary = salary + bonus
print("Total salary :",Total_salary)

difference = Total_salary - bonus
print("Difference :",difference)

multiply = Total_salary * 2
print("Multiply by salary :",multiply)

Division = Total_salary / 2
print("Division :",Division)
```

```python
#Take two numbers from the user and calculate
a = 100
b = 10
print("Addition:",a+b)
print("Multiplication :",a * b)
print("Subtraction :",a - b)
print("Division :",a/b)
print("Floor division :",a//b)
print("Power :",a**b)
print("Modulus :",a%b)
```

```python
#Comparison operators 
age = 22
print("Age is equal to 22:", age == 22)
print("Age is greater than 18 :",age > 18)
print("Age is less than 25 :",age < 25)
print("Age is not equal to 30 :",age != 30)
print("Age is greater than and equal to 21 :",age >= 21)
print("Age is Lessthan and equal to 21 :",age <= 21)
```

```python
#Assignment operators 
salary = 30000

#Increase salary by 5000
salary += 5000
print("Increase salary :",salary)

#Decrease salary by 2000
salary -= 2000
print("New salary :",salary)

#Multiply salary by 2
salary *= 2
print("Multiply :",salary)

#Divide salary by 2
salary /= 2
print("Divide salary :",salary)
```

```python
#Logical operators 
age = 22
salary = 40000

chack1 = age > 18 and salary > 30000
print("Chack1 ans :",chack1)

chack2 = age > 25 or salary > 25000
print("Chack2 ans :",chack2)

chack3 = age >23 and salary < 30000
print("Chack3 ans :",chack3)

chack4 = age > 24 or salary < 30000
print("Chack4 ans :",chack4)
```
