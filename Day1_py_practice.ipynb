```python
#A user enters age a string convert into integer and add 5 years
age = "22"
print("Old Age :",age,type(age))

new_age = int(age) + 5
print("New_Age :",new_age,type(new_age))
```

```python
#Convert the int number into float 
num = 50
print("Num :",num,type(num))

convert_f = float(num)
print("Convert :",convert_f,type(convert_f))
```
