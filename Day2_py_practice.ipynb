#Topic Operators and input

```python
#Take two number from the user and use all arithmetic oprators 
x = int(input("Enter first number :"))
y = int(input("Enter secound number :"))
print("Addition :", x + y)
print("Subtraction :",x - y)
print("Multiplication :",x * y)
print("Division :", x / y)
print("Floar division :", x // y)
print("Power :",x ** y)
```

```python
#Take number from the user and find square and cube 
x = int(input("Enter number :"))
print("Square :", x * x)
print("Cube :", x**3)
```

```python
#Take marks of 3 subject from the user and calculate average 
s1 = int(input("Maths :"))
s2 = int(input("Science :"))
s3 = int(input("English :"))
Total = s1 + s2 + s3
print("Total marks :",Total)
print("Average :",Total / 3)
```

```python
#Compare two number 
x = 89
y = 99

print("Greater than :", x > y)
print("Less than :", x < y)
print("Equal to :", x == y)
```

```python
#Take user age as input and chack person is eligible for voting 
age = int(input("Enter your age :"))
print("Eligible / not eligible :",age >= 18)
```

```python
#Logical operator 
marks = int(input("Enter marks :"))
Attendance = int(input("Enter Attendance :"))
print("Student pass / Fail :",marks >= 40 and Attendance >= 75)
```

```python
#Login system 
x = input("Enter user name :")
y = int(input("Enter password :"))
if x == "admin" and y == 1234:
    print("Login Successful")
else:
    print("Invalid Login")
```

```python
#Discount calculator 
product_price = int(input("Enter product price :"))
discount_total = product_price * 0.1
discount = product_price - discount_total

if product_price >= 5000:
    print("Total discount :",discount_total)
    print("Total Bill with discount :",discount)
else :
    print("Total Bill :",product_price)
```

#Assignment Operators 

```python
#Create variable
balance = 1000

#Add 500 
new_balance = balance + 500
print("Add :",new_balance)

#Subtract 200
new_balance -= 200
print("Subtract :",new_balance)

#Multiply by 2
new_balance *= 2
print("Multiply :",new_balance)
```
